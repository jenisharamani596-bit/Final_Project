```python
# Country Behavior Summary
country_data = df.groupby('Country').agg({
    'Year': 'count',
    'Happiness_Score': 'mean'
}).rename(columns={'Year': 'RecordCount', 'Happiness_Score': 'AvgHappiness'})

country_data
```

```python
# Manual classification
def classify_country(row):
    if row['AvgHappiness'] >= 5.6 and row['RecordCount'] >= 90:
        return 'High Happiness'
    elif row['AvgHappiness'] >= 5.4:
        return 'Medium Happiness'
    else:
        return 'Low Happiness'

# Apply the function to classify countries
country_data['Category'] = country_data.apply(classify_country, axis=1)

# Count the number of countries in each category
category_counts = country_data['Category'].value_counts()

category_counts
```

```python
country_data
```

```python
import matplotlib.pyplot as plt

# Scatter plot
plt.figure(figsize=(10, 6))

for label, group in country_data.groupby('Category'):
    print(f'{label} countries: {len(group)}')
    plt.scatter(group['RecordCount'], group['AvgHappiness'], label=label, s=100)

    for country, row in group.iterrows():
        plt.annotate(country, (row['RecordCount'], row['AvgHappiness']),
                     xytext=(5, 5), textcoords='offset points')

plt.xlabel('Record Count')
plt.ylabel('Average Happiness Score')
plt.title('Country Happiness Segmentation')
plt.legend()
plt.savefig('scatter_plot.png', dpi=300, bbox_inches='tight')

plt.show()
```
